```python
import pandas as pd

csv_path = "youtube.csv"
df = pd.read_csv(csv_path)

print(df.columns.tolist())
```

```python
keep_cols = {
    'comment_id': 'id',
    'video_id': 'video id',
    'author': 'author',
    'comment_text': 'text',
    'votes': 'likeCount',
    'comment_published_at_approx': 'created_time',
    'video_date': 'video_date',
}

df = df[list(keep_cols.keys())].rename(columns=keep_cols)
```

```python
output_csv_path = csv_path.replace(".csv", "_cleaned.csv")
df.to_csv(output_csv_path, index=False)
print(f"Exported to {output_csv_path}")
```

```python
null_pct = (df.isna().mean() * 100).round(2).sort_values(ascending=False)
print(null_pct)
```
